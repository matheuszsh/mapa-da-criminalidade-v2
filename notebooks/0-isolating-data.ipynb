### Dependências

```python
import pandas as pd
```

#### Importando Dataset

```python
try:
    dataset = pd.read_excel('../data/SPDadosCriminais_2025.xlsx')
except:
    print("Dataset não encontrado. Execute o script no caminho: /scripts/download_dataset.py")
```

#### Filtando por dados referentes ao município de São Paulo

```python
# Limpeza para gerar a planilha SP_CAPITAL
sao_paulo_capital = dataset.loc[(dataset['NOME_MUNICIPIO'].isin(['S.PAULO'])) & dataset['NOME_MUNICIPIO_CIRCUNSCRIÇÃO'].isin(['S.PAULO'])]

sao_paulo_capital.to_excel('../data/SP_capital_dadosCriminais_2025.xlsx')

```
